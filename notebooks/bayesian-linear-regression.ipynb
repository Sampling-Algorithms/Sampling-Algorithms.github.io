# Bayesian Linear Regression (Notebook)
A minimal notebook to sample posterior parameters for a linear model.

```python
import numpy as np
import pymc as pm
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 100)
alpha_true, beta_true, sigma_true = 1.0, 2.5, 2.0
y = alpha_true + beta_true * x + rng.normal(0, sigma_true, size=x.size)
```

```python
with pm.Model() as model:
    alpha = pm.Normal('alpha', mu=0, sigma=10)
    beta = pm.Normal('beta', mu=0, sigma=10)
    sigma = pm.HalfNormal('sigma', sigma=2)

    mu = alpha + beta * x
    pm.Normal('obs', mu=mu, sigma=sigma, observed=y)

    idata = pm.sample(1000, tune=1000, chains=4, target_accept=0.9, random_seed=42)
```

```python
pm.plot_trace(idata, var_names=['alpha', 'beta', 'sigma']);
plt.tight_layout()
```
